```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Library untuk Machine Learning
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from imblearn.over_sampling import SMOTE
from ucimlrepo import fetch_ucirepo

# Memuat dataset
print("Memuat dataset...")
dataset = fetch_ucirepo(id=350)
df = pd.concat([dataset.data.features, dataset.data.targets], axis=1)

# Jika ada kolom ID atau Unnamed yang terbawa, kita hapus
if 'ID' in df.columns:
    df = df.drop(columns=['ID'])

print("Dataset berhasil dimuat! Dimensi data:", df.shape)
df.head()
```

```python
# ---------------------------------------------------------
# 1. RINGKASAN STATISTIK DESKRIPTIF
# ---------------------------------------------------------
print("=== RINGKASAN STATISTIK DESKRIPTIF ===")
display(df.describe().T)


# ---------------------------------------------------------
# 2. VISUALISASI DISTRIBUSI TARGET (Y)
# ---------------------------------------------------------
target_col = dataset.data.targets.columns[0]

plt.figure(figsize=(7, 5))
ax = sns.countplot(data=df, x=target_col, palette="viridis", hue=target_col, legend=False)

plt.title(f"Distribusi Kelas Target ({target_col})", fontsize=14, fontweight="bold", pad=15)
plt.xlabel("Status Default Payment (0 = Tidak, 1 = Ya)", fontsize=11)
plt.ylabel("Jumlah Nasabah", fontsize=11)

total_data = len(df)
for p in ax.patches:
    count = int(p.get_height())
    percentage = f"{100 * count / total_data:.1f}%"
    ax.annotate(
        f"{count:,}\n({percentage})",
        (p.get_x() + p.get_width() / 2.0, p.get_height()),
        ha="center",
        va="bottom",
        fontsize=10,
        fontweight="bold",
        color="black"
    )

plt.ylim(0, df[target_col].value_counts().max() * 1.15)
plt.tight_layout()
plt.show()


# ---------------------------------------------------------
# 3. HEATMAP KORELASI ANTAR FITUR
# ---------------------------------------------------------
plt.figure(figsize=(18, 14))

# Menghitung matriks korelasi
corr_matrix = df.corr()


mask = np.triu(np.ones_like(corr_matrix, dtype=bool))

sns.heatmap(
    corr_matrix,
    mask=mask,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    vmin=-1, vmax=1,
    linewidths=0.5,
    cbar_kws={"shrink": 0.8},
    annot_kws={"size": 8}
)

plt.title("Heatmap Korelasi Antar Fitur & Target", fontsize=16, fontweight="bold", pad=20)
plt.xticks(rotation=45, ha="right")
plt.yticks(rotation=0)
plt.tight_layout()
plt.show()
```
